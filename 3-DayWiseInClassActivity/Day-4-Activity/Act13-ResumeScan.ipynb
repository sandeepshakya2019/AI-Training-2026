### Named Entity Recognization and POS Tagging

```python
import spacy

# Load spaCy's lightweight English model
nlp = spacy.load("en_core_web_sm")

# Sample unstructured text input
raw_resume_text = """
Rahul Sharma is a Software Developer living in New Delhi.
He worked at Google for 3 years starting from March 2021.
Rahul graduated from Delhi University and has expertise
in Python, Machine Learning and AWS.
"""

# Process text
print("Processing text with spaCy...\n")

doc = nlp(raw_resume_text)

# Extract Named Entities
print("--- Extracted Named Entities ---")

for ent in doc.ents:
    print(f"Entity: {ent.text:<20} | Type: {ent.label_:<10} "
          f"({spacy.explain(ent.label_)})")

# Extract key nouns and skills
print("\n--- Identified Keywords/Nouns ---")

nouns = [
    token.text for token in doc
    if token.pos_ in ["NOUN", "PROPN"]
]

print("Key Nouns Found:", list(set(nouns)))
```

```python
from spacy import displacy

displacy.render(doc, style="ent", jupyter=True)
```
