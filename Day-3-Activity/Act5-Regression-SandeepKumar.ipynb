### Activity 5 Supervised Regression

```python
# import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
X_pages = [[10], [20], [30], [40]] #2d format
y_days = [2, 4, 6, 8] #1d format
```

```python
sns.scatterplot(x=[i[0] for i in X_pages], y=y_days, linewidth=2, marker="o", color="blue", label="Data Points")
plt.xlabel("No of pages")
plt.ylabel("No of days")
plt.legend()
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
model.fit(X_pages, y_days)
```

```python
model.get_params()
```

```python
print(model.coef_)
print(model.intercept_)
```

```python
print("No of pages: 50")
print("Predicted days:", model.predict([[50]]))
```

```python
sns.scatterplot(
    x=[i[0] for i in X_pages],
    y=y_days,
    color="blue"
)

plt.plot(
    [i[0] for i in X_pages],
    model.predict(X_pages),
    linestyle="--",
    color="red",
    linewidth=2,
    label="Regression Line"
)

plt.xlabel("Pages")
plt.ylabel("Days")
plt.legend()
plt.show()
```

```python
m = model.coef_[0]
c = model.intercept_

x = np.array([i[0] for i in X_pages])
y = m * x + c

sns.scatterplot(x=x, y=y_days, color="blue")
plt.plot(x, y, linestyle="--", color="red")

plt.xlabel("Pages")
plt.ylabel("Days")
plt.show()
```
