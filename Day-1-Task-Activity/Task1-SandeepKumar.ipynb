Q1. Write a function. Given list of strings, find unique values and return them

```python
def unique_values(l):
    result = []
    for i in l:
        if i not in result:
            result.append(i)
    return result

l = ['A', 'B', 'A', 'C', 'B', 'D']
unique_values(l)
```

Q2. Write a function. Given list of strings, convert every string to a number using following encoding

Every string is given a number based on its alphabetical order in given list
return the converted list

Ex. l=['A','B','A','D']

'A' given 0
'B' given 1
'D' given 2

O/p return value of function [0,1,0,2]

```python
def encode(l):
    unique = sorted(set(l))
    result = []
    for i in l:
        result.append(unique.index(i))
    return result

l = ['A', 'B', 'A', 'D']
encode(l)
```

Q3. Write a function. Given list of numbers 
find mean, median and mode without using any library.

```python
def statistics(l):
    n = len(l)
    mean = sum(l) / n
    l = sorted(l)
    if n % 2 == 0:
        median = (l[n//2 - 1] + l[n//2]) / 2
    else:
        median = l[n//2]
    mode = max(l, key=l.count)

    return mean, median, mode


l = [1, 2, 3, 3, 4, 5]

mean, median, mode = statistics(l)

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
```
